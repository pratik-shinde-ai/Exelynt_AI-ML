```python
#abstractmethod
from abc import ABC , abstractmethod
class Vehicle(ABC):
    
    @abstractmethod
    def drive(self):
        print("Please wear the set built")

class Drive_motar(Vehicle):
    def drive(self):
        print("drive the car")
    
    def speed(self):
        print("avoid the speed more than 60")


obj=Drive_motar()
obj.drive()
obj.speed()

```

```python
# inheritance
class Animal():
    def eat(self):
        print("I can Eat")

class Dog(Animal):
    def speak(self):
        print("woof")
    
obj=Dog()
obj.eat()
obj.speak()
     

```

```python
#Encapsulation
class Info():
    def f_name(self,name):
        self.__name="" # here we make the pravate variable

    def get_method(self,name):
        self.__name=name
    
    def set_method(self):
        return self.__name

obj=Info()
obj.get_method("Pratik")
obj.set_method()
```
