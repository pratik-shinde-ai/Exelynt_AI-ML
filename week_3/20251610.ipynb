```python
class Animal:
    def speak(self):
        print("I can speak")

class Dog():
    def speak(self):
        print("woof")

class Cat():
    def speak(self):
        print("maue")

a=Animal()
d=Dog()
c=Cat()

d.speak()
c.speak()
```

```python
#polymorphism
class Father():
    def speak(self):
        print("I Can speak")
class Son(Father):
    def eat(self):
        print("I can Eat")


s=Son()
s.speak()
s.eat()

```
