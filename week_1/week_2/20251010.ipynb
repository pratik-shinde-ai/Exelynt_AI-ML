```python
#inheritance- copy all property in child class
class Animal:
    def eat(self):
        print("i can eat") 

class Dog(Animal):
    def speak(self):
        print("My voice is boof")

d=Dog()
d.eat()
d.speak()
```

```python
#polymorphism- many forms
class Animal:
    def speak(self):
        print("I can speak")

class Dog:
    def speak(self):
        print("woof")

class Cat:
    def speak(self):
        print("maue")

a=Animal()
d=Dog()
c=Cat()
d.speak()
c.speak()
```

```python
#Encapsulation- bind the data we achieve the encapsulation  using access modifiers and get and set method

class A:
    def my_name(self,name):
        self.__name="" #pravate variable
    
    def set_name(self,name):
        self.__name=name
    
    def get_name(self):
        return self.__name

# we print the name outside the class  
obj=A()
obj.set_name("pratik")
print(obj.get_name())

        
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        print("for starting check the car condition")

class Car(Vehicle):
    def start(self):
        print("Car starts with a key")

c = Car()
c.start()
```
