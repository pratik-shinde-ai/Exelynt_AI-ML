```python
#add disipline and remove in-disipline mark
class Examscore:
    def __init__(self,name,exam_score):
        self.name=name
        self.exam_score=exam_score

    def disiplinemark(self, extra_score):
        self.exam_score +=extra_score
        print(f"the extra mark is {extra_score} after the extra makr total score is {self.exam_score} ")
        
    def indisiplinemark(self, lose_mark):
        self.exam_score -=lose_mark
        print(f"the lose mark becuse of indisipline is {lose_mark} and total score became {self.exam_score}")

obj=Examscore("Pratik",50)
obj.disiplinemark(20)
obj.indisiplinemark(10)
        
        
    

```
