# Encapsulation
it achiever thorow using get and set method for that we need in a class pravate variable

```python
class A:
    def __init__(self):
        self.__name=""    # heere we make the pravate variabel

#now we want print above variabel in a outside of the class for that we use get and set method
    
    #set name
    def set_name(self,name):
        self.__name=name
    
    #get name
    def get_name(self):
        return self.__name

#create the object outside of the class and access the name  
obj=A()
obj.set_name("Pratik")
print(obj.get_name())
```

# Inheritance
Inherite or copy or grap the property from parent class to child class 

```python
class Animal:
    def eat(self):
        print("i can eat")


class Dog(Animal):   # here we make the dog child class and inheritance from parent class 
    def bark(self):
        print("I Can Bark")


d=Dog()
d.eat() # this eat method is calling from animal class it means parents class 
d.bark()
```

# Abstraction
Hide the data and showing only nesseary details ex- mobile phones only look app feature we not look behinde the logic or process

```python
from abc import ABC , abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def drive(self):
        pass

class Car(Vehicle):
    def drive(self):
        print("I can Drive")

c=Car()
c.drive()
```
