```python
#polymorphism
class Animal():
    def speak(self):
        print("i can speak")

class Dog():
    def speak(self):
        print("woof")

a=Animal()
d=Dog()
a.speak()
d.speak()
```

```python
#inheritance
class Animal():
    def speak(self):
        print("i can speak")

class Dog(Animal):
    def eat(self):
        print("I can Eat")

d=Dog()
d.speak()
d.eat()


```

```python
#abstraction 
from abc import ABC, abstractmethod
class Animal():
    @abstractmethod
    def speak():
        pass

class Dog():
    def speak(self):
        print("I can Speak")
    
    def eat(self):
        print("I can Eat")

d=Dog()
d.speak()
d.eat()
```
